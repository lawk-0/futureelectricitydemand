This notebook inspects the PJM Hourly Energy Consumption dataset, checks time regularity, identifies missing and duplicate timestamps, and analyzes diurnal, weekly, and annual seasonality.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from src.data.loader import DataLoader
from src.data.validation import DataValidator

loader = DataLoader()
df = loader.load_dataset('data/raw/PJME_hourly.csv')
print('Dataset Shape:', df.shape)
df.head()
```

```python
validator = DataValidator()
audit = validator.validate(df)
print('Data Integrity Score:', audit['integrity_score'], '%')
print('Duplicate Timestamps:', audit['duplicate_timestamps_count'])
print('Missing Grid Timestamps:', audit['missing_timestamps_count'])
```

```python
# Diurnal Hourly Pattern
df['hour'] = df['timestamp'].dt.hour
hourly_mean = df.groupby('hour')['consumption'].mean()
plt.figure(figsize=(10, 4))
hourly_mean.plot(kind='line', marker='o', color='#1f77b4')
plt.title('Diurnal Electricity Demand Profile - Mean MW by Hour')
plt.xlabel('Hour of Day (0-23)')
plt.ylabel('Consumption (MW)')
plt.grid(True, alpha=0.3)
plt.show()
```
