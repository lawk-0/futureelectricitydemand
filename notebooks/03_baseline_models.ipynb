Evaluates Seasonal Naive benchmarks (24h daily and 168h weekly lags) and Facebook Prophet Generalized Additive Models.

```python
import pandas as pd
from src.models.baseline import SeasonalNaiveModel
from src.models.prophet_model import ProphetDemandModel
from src.evaluation.metrics import evaluate_forecast

df = pd.read_csv('data/processed/clean_consumption.csv')
df['timestamp'] = pd.to_datetime(df['timestamp'])

train_df = df.iloc[-1000:-168]
test_df = df.iloc[-168:]

naive = SeasonalNaiveModel(season_lag=168).fit(train_df)
forecast_naive = naive.predict(horizon=168, start_timestamp=test_df['timestamp'].iloc[0])
metrics_naive = evaluate_forecast(test_df['consumption'], forecast_naive['predicted'])
print('Seasonal Naive 168h MAE:', metrics_naive['MAE'], 'MW')
print('Seasonal Naive 168h RMSE:', metrics_naive['RMSE'], 'MW')
```
